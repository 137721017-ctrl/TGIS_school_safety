```python
import csv
from dbfread import DBF

def dbf_to_csv(dbf_file_path, csv_file_path):
    """
    将DBF文件转换为CSV格式
    :param dbf_file_path: 输入DBF文件的路径
    :param csv_file_path: 输出CSV文件的路径
    """
    try:
        # 读取DBF文件
        dbf_table = DBF(dbf_file_path, encoding='utf-8')
        
        # 创建并写入CSV文件
        with open(csv_file_path, 'w', newline='', encoding='utf-8') as csvfile:
            writer = csv.writer(csvfile)
            
            # 写入列名（表头）
            writer.writerow(dbf_table.field_names)
            
            # 逐行写入数据记录
            for record in dbf_table:
                writer.writerow([record[field_name] for field_name in dbf_table.field_names])
                
        print(f"转换成功！CSV文件已保存至: {csv_file_path}")
        
    except Exception as e:
        print(f"处理过程中发生错误: {str(e)}")

# 使用示例
if __name__ == "__main__":
    input_dbf_path = 'E:\\SchoolPoints_APS_2024_08_28\\SchoolPoints_APS_2024_08_28.dbf'  # 请替换为您的DBF文件实际路径
    output_csv_path = 'E:\\newyorkschool.csv'  # 指定输出的CSV文件路径
    dbf_to_csv(input_dbf_path, output_csv_path)
```

```python
import pandas as pd
import chardet

def detect_file_encoding(file_path):
    """
    自动检测文件编码格式
    """
    with open(file_path, 'rb') as f:
        raw_data = f.read(10000)  # 读取文件前10000字节进行编码检测
        result = chardet.detect(raw_data)
        return result['encoding']

def safe_read_csv(file_path, encodings=None):
    """
    安全读取CSV文件，尝试多种编码格式
    """
    if encodings is None:
        encodings = ['utf-8', 'gbk', 'gb2312', 'latin-1', 'iso-8859-1']
    
    # 首先尝试自动检测编码
    try:
        detected_encoding = detect_file_encoding(file_path)
        if detected_encoding:
            encodings.insert(0, detected_encoding)
    except:
        pass
    
    # 尝试各种编码格式读取文件
    for encoding in encodings:
        try:
            df = pd.read_csv(file_path, encoding=encoding)
            print(f"成功使用编码: {encoding}")
            return df
        except UnicodeDecodeError:
            continue
        except Exception as e:
            print(f"编码 {encoding} 尝试失败: {str(e)}")
            continue
    
    raise ValueError("无法找到合适的编码格式读取文件")

# 主处理程序
try:
    # 安全读取CSV文件
    file_path = 'E:\\newyorkschool.csv'
    df = safe_read_csv(file_path)
    
    # 指定需要统计的列名
    target_column = 'Location_Category_Description'
    
    # 检查列是否存在
    if target_column not in df.columns:
        available_columns = ', '.join(df.columns)
        raise ValueError(f"列 '{target_column}' 不存在。可用列名: {available_columns}")
    
    # 计算该列各值的出现次数
    values_counts = df[target_column].value_counts()
    
    # 打印统计结果
    print(f"统计列 '{target_column}' 的频次结果:")
    print(values_counts)
    
    # 保存结果到CSV文件（保留索引以显示统计值）
    output_path = 'E:\\newyorkschoolcategory.csv'
    values_counts.to_csv(output_path, index=True)
    print(f"结果已保存到: {output_path}")
    
except FileNotFoundError:
    print(f"错误: 文件 {file_path} 不存在，请检查文件路径")
except Exception as e:
    print(f"处理过程中发生错误: {str(e)}")
```

```python
import os
from dbfread import DBF  # 需要先安装dbfread库

# 定义源文件夹和目标存储位置
source_dir = r"E:\\newyork_arrests"
output_dir = "E:\\"  # 直接存储在E盘根目录

# 创建输出目录（如果不存在）
os.makedirs(output_dir, exist_ok=True)

# 处理2006到2024年的所有文件
for year in range(2006, 2025):  # 2025不包括，所以实际到2024年
    # 构建文件名
    dbf_filename = f"{year}_buff_300.dbf"
    csv_filename = f"{year}_buff_300.csv"  # CSV文件名
    
    # 构建完整路径
    dbf_path = os.path.join(source_dir, dbf_filename)
    csv_path = os.path.join(output_dir, csv_filename)
    
    print(f"正在转换: {dbf_filename} -> {csv_filename}")
    
    try:
        # 读取DBF文件
        dbf_table = DBF(dbf_path, encoding='utf-8')
        
        # 写入CSV文件
        with open(csv_path, 'w', encoding='utf-8') as csvfile:
            # 写入表头
            header = ','.join(dbf_table.field_names)
            csvfile.write(header + '\n')
            
            # 写入所有记录
            for record in dbf_table:
                line = ','.join(str(record[field]) for field in dbf_table.field_names)
                csvfile.write(line + '\n')
        
        print(f"成功转换: {csv_filename}")
    
    except Exception as e:
        print(f"转换失败 ({dbf_filename}): {str(e)}")

print("所有文件转换完成！")
```

```python
import pandas as pd
import glob
import os
from tqdm import tqdm  # 进度条库，需要安装

# 设置文件夹路径和输出路径
input_folder = r"E:\\newyork_arrests_300_year"
output_file = r"E:\\crimes_frequency_summary.csv"

# 创建tqdm的notebook版本
try:
    from tqdm.notebook import tqdm
except:
    from tqdm import tqdm

# 确保文件夹存在
if not os.path.exists(input_folder):
    raise ValueError(f"输入文件夹不存在: {input_folder}")

# 获取所有年份的CSV文件路径
csv_files = glob.glob(os.path.join(input_folder, "*_buff_300.csv"))
if not csv_files:
    raise ValueError(f"在 {input_folder} 中没有找到CSV文件")

print(f"找到 {len(csv_files)} 个文件待处理...")

# 初始化一个空的DataFrame用于存储所有频率
all_frequencies = pd.DataFrame()

# 设置进度条
progress_bar = tqdm(csv_files, desc="处理文件中")

# 处理每个年份的文件
for file_path in progress_bar:
    # 从文件名中提取年份
    file_name = os.path.basename(file_path)
    year = file_name.split("_")[0]
    
    # 更新进度条描述
    progress_bar.set_description(f"处理 {year} 年数据")
    
    try:
        # 读取CSV文件（只读取需要的列以提高效率）
        df = pd.read_csv(file_path, usecols=['PD_DESC'])
        
        # 计算每种罪行的频率
        frequency = df['PD_DESC'].value_counts().reset_index()
        frequency.columns = ['Crime', year]  # 设置列名（Crime和年份）
        
        # 合并到汇总表
        if all_frequencies.empty:
            all_frequencies = frequency
        else:
            all_frequencies = pd.merge(all_frequencies, frequency, on='Crime', how='outer')
    
    except Exception as e:
        print(f"\n处理文件 {file_name} 时出错: {str(e)}")

# 处理缺失值（用0填充）
all_frequencies = all_frequencies.fillna(0)

# 将年度列转换为整数（因为合并时可能变为浮点数）
year_columns = [col for col in all_frequencies.columns if col != 'Crime']
for col in year_columns:
    all_frequencies[col] = all_frequencies[col].astype(int)

# 按罪行名称排序
all_frequencies = all_frequencies.sort_values(by='Crime')

# 将年份列按顺序排列
year_columns_sorted = sorted(year_columns)
sorted_columns = ['Crime'] + year_columns_sorted
all_frequencies = all_frequencies[sorted_columns]

# 保存汇总结果
all_frequencies.to_csv(output_file, index=False)

print("\n处理完成!")
print(f"共统计了 {len(all_frequencies)} 种不同罪行")
print(f"结果已保存至: {output_file}")

# 显示前5行结果预览
all_frequencies.head()
```

```python
import pandas as pd
import glob
import os

# 设置文件夹路径和输出路径
input_folder = r"E:\newyork_arrests_300_year"
output_file = r"E:\crimes_frequency_summary.csv"

# 确保文件夹存在
if not os.path.exists(input_folder):
    raise ValueError(f"输入文件夹不存在: {input_folder}")

# 获取所有年份的CSV文件路径
csv_files = glob.glob(os.path.join(input_folder, "*_buff_300.csv"))
if not csv_files:
    raise ValueError(f"在 {input_folder} 中没有找到CSV文件")

print(f"找到 {len(csv_files)} 个文件待处理...")

# 初始化一个空的DataFrame用于存储所有频率
all_frequencies = pd.DataFrame()

# 处理每个年份的文件
for file_path in csv_files:
    # 从文件名中提取年份
    file_name = os.path.basename(file_path)
    year = file_name.split("_")[0]
    
    print(f"处理 {year} 年数据...")
    
    try:
        # 检查文件大小
        file_size = os.path.getsize(file_path)
        print(f"  文件大小: {file_size/(1024*1024):.2f} MB")
        
        # 读取CSV文件
        df = pd.read_csv(file_path, usecols=['OFNS_DESC'])
        
        # 计算每种罪行的频率
        frequency = df['OFNS_DESC'].value_counts().reset_index()
        frequency.columns = ['Crime', year]  # 设置列名（Crime和年份）
        
        # 合并到汇总表
        if all_frequencies.empty:
            all_frequencies = frequency
        else:
            all_frequencies = all_frequencies.merge(frequency, on='Crime', how='outer')
        
        print(f"  成功处理 {len(frequency)} 种罪行类型")
    
    except Exception as e:
        print(f"  处理文件 {file_name} 时出错: {str(e)}")
        continue

# 处理缺失值（用0填充）
all_frequencies = all_frequencies.fillna(0)

# 将年度列转换为整数（因为合并时可能变为浮点数）
year_columns = [col for col in all_frequencies.columns if col != 'Crime']
for col in year_columns:
    all_frequencies[col] = all_frequencies[col].astype(int)

# 按罪行名称排序
all_frequencies = all_frequencies.sort_values(by='Crime')

# 将年份列按顺序排列
year_columns_sorted = sorted(year_columns)
sorted_columns = ['Crime'] + year_columns_sorted
all_frequencies = all_frequencies[sorted_columns]

# 保存汇总结果
all_frequencies.to_csv(output_file, index=False)

print("\n所有文件处理完成!")
print(f"共统计了 {len(all_frequencies)} 种不同罪行")
print(f"结果已保存至: {output_file}")

# 显示前5行结果预览
all_frequencies.head()
```

```python
import pandas as pd

# 读取CSV文件
file_path = r'E:\\hongkong_accidents_300.csv'
df = pd.read_csv(file_path)

# 检查列是否存在
if not all(col in df.columns for col in ['Year', 'Type_of_Co']):
    missing = [col for col in ['Year', 'Type_of_Co'] if col not in df.columns]
    raise ValueError(f"列名不存在: {missing}")

# 创建频数统计表（交叉表）
frequency_table = pd.crosstab(
    index=df['Type_of_Co'],  # 行（转换为列）
    columns=df['Year'],      # 列（转换为行）
    margins=True,            # 添加总计行/列
    margins_name='Total'     # 总计项名称
)

# 保存结果到E盘
output_path = r'E:\\accidents_frequency_summary.csv'
frequency_table.to_csv(output_path, index=True)

print(f"统计表已保存至: {output_path}")
print("\n预览统计表:")
print(frequency_table.head())
```
